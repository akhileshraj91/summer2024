```python
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pandas as pd
```

```python

def get_data_dir(subfolder):
    current_dir = os.getcwd()
    print(current_dir)
    return os.path.join(current_dir, "experiment_data", f"{subfolder}")
```

```python
DATA_DIR = get_data_dir("data_generation_new")

csv_file_name = 'training_dataset.csv'
csv_file_path = os.path.join(DATA_DIR, csv_file_name)
data = pd.read_csv(csv_file_path)
df = pd.DataFrame(data)
# Assuming 'df' is your DataFrame with the relevant data
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# Plot each scatter plot
sns.scatterplot(data=df, x='TOT_INS_PER_CYC', y='Next_Progress', ax=ax[0])
ax[0].set_title('TOT_INS_PER_CYC vs progress')

sns.scatterplot(data=df, x='L3_TCM_PER_TCA', y='Next_Progress', ax=ax[1])
ax[1].set_title('TOT_L3_TCM_PER_TCA vs progress')

sns.scatterplot(data=df, x='TOT_STL_PER_CYC', y='Next_Progress', ax=ax[2])
ax[2].set_title('STL_PER_CYC vs progress')

plt.tight_layout()
plt.show()
```
