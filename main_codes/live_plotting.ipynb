```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import time

fig, ax = plt.subplots()
ax.set_xlabel('Iteration')
ax.set_ylabel('Loss')
ax.set_title('Live Loss Plot')

x_vals, y_vals = [], []
(line,) = ax.plot([], [], 'r-')

for i in range(100):
    x_vals.append(i)
    y_vals.append(np.exp(-i / 20) + np.random.rand() * 0.1)

    line.set_data(x_vals, y_vals)
    ax.relim()
    ax.autoscale_view()
    fig.canvas.draw_idle()  # Redraw the canvas

    plt.pause(0.05)  # Pause to allow GUI event loop to update
```
